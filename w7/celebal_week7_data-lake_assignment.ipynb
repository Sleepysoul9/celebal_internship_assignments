```python
import pandas as pd
import random
from faker import Faker
from datetime import datetime, timedelta
```

```python
fake = Faker("en_IN")

# Makes dataset reproducible
random.seed(42)
Faker.seed(42)
```

```python
cities = {
    "Delhi": "Delhi",
    "Mumbai": "Maharashtra",
    "Pune": "Maharashtra",
    "Jaipur": "Rajasthan",
    "Lucknow": "Uttar Pradesh",
    "Hyderabad": "Telangana",
    "Bangalore": "Karnataka",
    "Chennai": "Tamil Nadu",
    "Ahmedabad": "Gujarat",
    "Kolkata": "West Bengal",
    "Indore": "Madhya Pradesh",
    "Bhopal": "Madhya Pradesh",
    "Surat": "Gujarat",
    "Patna": "Bihar",
    "Noida": "Uttar Pradesh"
}

membership_levels = [
    "Silver",
    "Gold",
    "Platinum"
]

customer_status = [
    "Active",
    "Inactive"
]
```

```python
customers = []

for customer_id in range(1001, 1501):

    city = random.choice(list(cities.keys()))

    customer = {
        "customer_id": customer_id,
        "customer_name": fake.name(),
        "gender": random.choice(["Male", "Female"]),
        "city": city,
        "state": cities[city],
        "email": fake.email(),
        "phone": str(random.randint(6000000000, 9999999999)),
        "membership": random.choice(membership_levels),
        "annual_spend": random.randint(5000, 150000),
        "last_purchase_date": fake.date_between(
            start_date="-2y",
            end_date="today"
        ),
        "status": random.choice(customer_status)
    }

    customers.append(customer)
```

```python
master_df = pd.DataFrame(customers)

print("Number of Customers:", len(master_df))

master_df.head()
```

```python
# -----------------------------
# Introduce Dirty Data
# -----------------------------

# 1. Add 20 duplicate records
duplicate_rows = master_df.sample(n=20, random_state=42)

master_df = pd.concat([master_df, duplicate_rows], ignore_index=True)

# 2. Make 10 customer names NULL
name_null_index = random.sample(list(master_df.index), 10)
master_df.loc[name_null_index, "customer_name"] = None

# 3. Make 10 annual_spend values NULL
spend_null_index = random.sample(list(master_df.index), 10)
master_df.loc[spend_null_index, "annual_spend"] = None

# 4. Make 10 invalid emails
email_invalid_index = random.sample(list(master_df.index), 10)
master_df.loc[email_invalid_index, "email"] = "invalid_email"

print("Dirty data created successfully!")
print("Total Records :", len(master_df))
```

```python
print("Duplicate Rows Added :", len(master_df) - 500)

print("\nMissing Customer Names :", master_df["customer_name"].isnull().sum())

print("Missing Annual Spend :", master_df["annual_spend"].isnull().sum())

print("Invalid Emails :",
      (master_df["email"] == "invalid_email").sum())
```

```python
import random
from pyspark.sql import Row

random.seed(42)

first_names = [
    "Aarav","Vivaan","Aditya","Krishna","Arjun","Rohan","Rahul",
    "Priya","Ananya","Sneha","Neha","Pooja","Kavya","Diya"
]

last_names = [
    "Sharma","Verma","Patel","Gupta","Singh",
    "Mehta","Jain","Joshi","Kapoor","Agarwal"
]

cities = [
    "Delhi","Mumbai","Pune","Jaipur",
    "Ahmedabad","Bengaluru","Chennai","Hyderabad"
]

records = []

for i in range(1, 501):

    first = random.choice(first_names)
    last = random.choice(last_names)

    email = f"{first.lower()}.{last.lower()}{i}@gmail.com"

    # Introduce invalid emails
    if random.random() < 0.05:
        email = "invalid_email"

    # Introduce null emails
    if random.random() < 0.05:
        email = None

    phone = "9" + "".join(str(random.randint(0,9)) for _ in range(9))

    # Introduce null phone numbers
    if random.random() < 0.05:
        phone = None

    city = random.choice(cities)

    records.append(
        Row(
            customer_id=i,
            first_name=first,
            last_name=last,
            email=email,
            phone=phone,
            city=city
        )
    )

# Add duplicate rows intentionally
records.extend(records[:20])

master_df = spark.createDataFrame(records)

print("Total Rows:", master_df.count())

display(master_df.limit(10))
```

```python
from pyspark.sql.functions import col

# Load the Delta table
master_df = spark.table("workspace.default.customer_master")

# Clean the data
clean_df = (
    master_df
    .dropDuplicates(["customer_id"])
    .filter(col("email").isNotNull())
    .filter(col("email").rlike(r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"))
)

# Save cleaned data back as a new Delta table
clean_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.customer_master_clean")

print("Original rows:", master_df.count())
print("Cleaned rows :", clean_df.count())

display(clean_df.limit(10))
```

```python
import random
from pyspark.sql import Row

random.seed(123)

# Existing cleaned data
clean_df = spark.table("workspace.default.customer_master_clean")

# Get existing customer IDs
existing_ids = [r.customer_id for r in clean_df.select("customer_id").collect()]
update_ids = random.sample(existing_ids, 50)

cities = [
    "Delhi", "Mumbai", "Pune", "Jaipur",
    "Ahmedabad", "Bengaluru", "Chennai", "Hyderabad"
]

incremental_records = []

# 50 updates
for cid in update_ids:
    incremental_records.append(
        Row(
            customer_id=cid,
            first_name=None,
            last_name=None,
            email=None,
            phone="8" + "".join(str(random.randint(0,9)) for _ in range(9)),
            city=random.choice(cities)
        )
    )

# 50 new customers
start_id = max(existing_ids) + 1

for i in range(50):
    cid = start_id + i

    incremental_records.append(
        Row(
            customer_id=cid,
            first_name=f"New{cid}",
            last_name="Customer",
            email=f"new{cid}@gmail.com",
            phone="9" + "".join(str(random.randint(0,9)) for _ in range(9)),
            city=random.choice(cities)
        )
    )

incremental_df = spark.createDataFrame(incremental_records)

incremental_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.customer_incremental")

print("Incremental rows:", incremental_df.count())

display(incremental_df.limit(10))
```

```python
from delta.tables import DeltaTable
from pyspark.sql.functions import coalesce

# Load Delta table
delta_table = DeltaTable.forName(spark, "workspace.default.customer_master_clean")

# Source data
incremental_df = spark.table("workspace.default.customer_incremental")

# MERGE operation
(
    delta_table.alias("target")
    .merge(
        incremental_df.alias("source"),
        "target.customer_id = source.customer_id"
    )
    .whenMatchedUpdate(
        set={
            "first_name": "coalesce(source.first_name, target.first_name)",
            "last_name": "coalesce(source.last_name, target.last_name)",
            "email": "coalesce(source.email, target.email)",
            "phone": "coalesce(source.phone, target.phone)",
            "city": "coalesce(source.city, target.city)"
        }
    )
    .whenNotMatchedInsert(
        values={
            "customer_id": "source.customer_id",
            "first_name": "source.first_name",
            "last_name": "source.last_name",
            "email": "source.email",
            "phone": "source.phone",
            "city": "source.city"
        }
    )
    .execute()
)

print("MERGE completed successfully!")
```
